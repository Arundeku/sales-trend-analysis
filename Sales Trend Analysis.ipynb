```python
import pandas as pd

# Load your dataset
df = pd.read_csv('Online Sales Data.csv')  # Adjust path if needed
```

```python
print(df.columns)
```

```python
df.rename(columns={
    'Date': 'order_date',
    'Total Revenue': 'amount',
    'Transaction ID': 'order_id'
}, inplace=True)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
```

```python
monthly_summary = df.groupby(['order_year', 'order_month']).agg(
    total_revenue=('amount', 'sum'),
    order_volume=('order_id', 'nunique')
).reset_index()
```

```python
monthly_summary = monthly_summary.sort_values(by=['order_year', 'order_month'])
```

```python
print(monthly_summary)
```

```python
# Group by year, month, and region
grouped_by_time_region = df.groupby(['order_year', 'order_month', 'Region']).agg(
    total_revenue=('amount', 'sum'),
    order_volume=('order_id', 'nunique')
).reset_index().sort_values(by=['order_year', 'order_month', 'Region'])

print(grouped_by_time_region)
```

```python
# Group by year, month, and product category
grouped_by_time_category = df.groupby(['order_year', 'order_month', 'Product Category']).agg(
    total_revenue=('amount', 'sum'),
    total_units_sold=('Units Sold', 'sum')
).reset_index().sort_values(by=['order_year', 'order_month', 'Product Category'])

print(grouped_by_time_category)
```

```python
# Revenue trend by region (pivot table format)
revenue_trend_by_region = df.pivot_table(
    index=['order_year', 'order_month'],
    columns='Region',
    values='amount',
    aggfunc='sum'
).fillna(0)

print(revenue_trend_by_region)
```

```python
# Count of payment methods used per month
payment_trends = df.groupby(['order_year', 'order_month', 'Payment Method']).size().reset_index(name='payment_count')

print(payment_trends)
```
